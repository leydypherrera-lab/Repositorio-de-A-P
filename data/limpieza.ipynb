# Limpieza de datos — Vibra Musical
Lee `raw/base_datos_original/canciones_original.csv`, lo limpia y guarda `datos.csv`.

```python
import pandas as pd

df = pd.read_csv('raw/base_datos_original/canciones_original.csv')
print(df.shape)
df.head()
```

```python
# 1. Revisar problemas
print('Nulos:\n', df.isna().sum())
print('Duplicados:', df.duplicated(subset=['Title', 'Artist']).sum())
df['Genre'].unique()
```

```python
# 2. Renombrar columnas al español
df = df.rename(columns={'Title': 'titulo', 'Artist': 'artista', 'Genre': 'genero', 'Year': 'anio'})
```

```python
# 3. Quitar espacios y unificar mayúsculas
df['titulo'] = df['titulo'].str.strip()
df['artista'] = df['artista'].str.strip()
df['genero'] = df['genero'].str.strip().str.title().replace({'R&b': 'R&B'})
```

```python
# 4. Quitar duplicados y filas sin año
df = df.drop_duplicates(subset=['titulo', 'artista'])
df = df.dropna(subset=['anio'])
df['anio'] = df['anio'].astype(int)
```

```python
# 5. Crear columna década, ordenar y poner id
df['decada'] = (df['anio'] // 10 * 10).astype(str) + 's'
df = df.sort_values(['anio', 'titulo']).reset_index(drop=True)
df.insert(0, 'id', df.index + 1)
df.head()
```

```python
# 6. Guardar el resultado
df.to_csv('datos.csv', index=False, encoding='utf-8')
print(df.shape)
df['genero'].value_counts()
```
